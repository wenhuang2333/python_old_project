```python
import matplotlib.pyplot as plt
import numpy as np
x=np.linspace(-1,1,50)
y1=(1-x**2)**0.5
kk=input()

y2=eval(kk)
plt.figure(figsize=(8,6))
plt.plot(y1,x,label='shift',color='blue',linewidth=1.0,linestyle='--')
plt.plot(x,y2,label='ok')
#坐标轴的域
plt.xlim((-5,5))
plt.ylim((-5,5))
#坐标轴标注
plt.xlabel("i an x")
plt.ylabel("i am y")
new_ticks=np.linspace(-1,2,5)
print(new_ticks)
#修改坐标轴的标志
plt.xticks(new_ticks)
plt.yticks([-2,-1,2,4],[r'$yep\ bro$',r'$nice\ control$',r'$ok\alpha\ guys$',r'$none\ problem$'])
#修改坐标轴的位置
#gca='get current axis'
ax=plt.gca()
ax.spines['right'].set_color('none')
ax.spines['top'].set_color('none')#将上边框和下边框消除
ax.xaxis.set_ticks_position('bottom')
ax.yaxis.set_ticks_position('left')#用下边来代替x轴，左边坐标轴代替y轴
ax.spines['bottom'].set_position(('data',0))#获取当前坐标系中底部坐标轴（即x轴），将底部坐标轴位置设置为数据坐标系下的位置0，就是将x轴的原点移动到y=0的位置，‘data’表示坐标轴的位置是根据数据坐标来设置的
ax.spines['left'].set_position(('data',0))#获取当前坐标系中的左侧坐标轴（即y轴），将左侧坐标轴的位置设置为在数据坐标系下的位置0.具体来说就是将y轴的远点移动到x=0的位置。
#默认情况下，坐标轴的交点可能不在（0，0），而是偏移的。通过这两行代码可以让x轴与y轴在（0，0）处交汇
#data或许可以换为outward,axes
#图例
plt.legend()
#与前文的lable对应
#plt.legend(handles=[y1,](加图例处理对象的限制，逗号不能少，下同）,labels=['aaa',](重命名),loc='best upper right...')
#添加注解v
x0=1
y0=2*x0+1
plt.scatter(x0,y0,s=50,color='r')#print出来一系列散点，s是size
plt.plot([x0,x0],[y0,y0],'k--',lw=1.5)#画出垂直于坐标轴的线，k代表黑色--代表虚线,lw是线粗
#注释呈现方法一：
plt.annotate(r"$2x+1=%s$"% y0,xy=(x0,y0),xycoords='data',xytext=(+30,-30),textcoords='offset points',fontsize=16,arrowprops=dict(arrowstyle='->',connectionstyle='arc3,rad=.2'))
#从左往右分别是正则表达式、指定注释应该放置的位置(x0,y0)坐标点，指定了注释位置xy参数使用的坐标系统（‘data’表示使用的是图标的数据坐标系），指定了文本注释相对于xy参数的指定位置的偏移量，指定了xytext使用的坐标系统为以点为单位的偏移量，设置了注释文本的字体大小为16，参数用于设置注释的箭头的属性，'->'指定了箭头样式，'arc3'指定了箭头连接点和文本的曲线样式，0.2设置了曲线的半径
#注释呈现方法二：
plt.text(-3.7,3,r'$This is the something \sigma_i \alpha_t$',fontdict=({'size':16,'color':'r'}))
plt.show()



```
